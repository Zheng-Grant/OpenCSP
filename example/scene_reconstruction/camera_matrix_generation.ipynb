This code is used for converting photomodeler camera parameters to OpenCSP format

```python
import numpy as np
import h5py

focal_length = 23.9066  # photomoderler number
format_size_w = 35.9857  # photomodelr number
format_size_h = 24.0  # photomodelr number
principal_x = 18.0397   # photomodeler number
principal_y = 12.0752   # photomodeler_number
image_size_w = 9504 # pixels, get from photo properties
image_size_h = 6336  # pixels, get from photo properties

# Assuming the principal point is at the center of the image

# distortion coefficients
k1 = 7.800e-05
k2 = -1.792e-07
p1 = -1.069e-05
p2 = -3.859e-06
distortion_coeff = np.array([k1, k2, p1, p2])

scale_factor_w = image_size_w / format_size_w
scale_factor_h = image_size_h / format_size_h

focal_length_w = focal_length * scale_factor_w
focal_length_h = focal_length * scale_factor_h

image_center_w = principal_x * scale_factor_w
image_center_h = principal_y * scale_factor_h

intrinsic_mat = np.array([[focal_length_w, 0, image_center_w],
                          [0, focal_length_h, image_center_h],
                          [0, 0, 1]], dtype=np.float32)

image_shape_xy = np.array([image_size_w, image_size_h])
name = 'replace this '

# Save the camera parameters to an HDF5 file
with h5py.File('camera.h5', 'w') as f:
    f.create_dataset('Camera/intrinsic_mat', data=intrinsic_mat)
    f.create_dataset('Camera/distortion_coef', data=distortion_coeff)
    f.create_dataset('Camera/image_shape_xy', data=image_shape_xy)
    f.create_dataset('Camera/name', data = name)









```
