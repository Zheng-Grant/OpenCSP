This code is used for converting photomodeler camera parameters to OpenCSP format

```python
import numpy as np
import h5py

focal_length = 20  # photomoderler number
format_size_w = 36  # photomodelr number
format_size_h = 24  # photomodelr number
image_size_w = 8640 # pixels
image_size_h = 5760  # pixels

# Assuming the principal point is at the center of the image

# distortion coefficients
k1 = -0.05576391
k2 = 0.03966674
p1 = -0.00055473
p2 = 0.00049181
distortion_coeff = np.array([k1, k2, p1, p2])

scale_factor_w = format_size_w / image_size_w
scale_factor_h = format_size_h / image_size_h

focal_length_w = focal_length * scale_factor_w
focal_length_h = focal_length * scale_factor_h

image_center_w = image_size_w / 2
image_center_h = image_size_h / 2

intrinsic_mat = np.array([[focal_length_w, 0, image_center_w],
                          [0, focal_length_h, image_center_h],
                          [0, 0, 1]], dtype=np.float32)

image_shape_xy = np.array([image_size_w, image_size_h])
name = 'Sony Alpha'

# Save the camera parameters to an HDF5 file
with h5py.File('camera_parameters.h5', 'w') as f:
    f.create_dataset('intrinsic_mat', data=intrinsic_mat)
    f.create_dataset('distortion_coef', data=distortion_coeff)
    f.create_dataset('image_shape_xy', data=image_shape_xy)
    f.attrs['name'] = name









```
